```python
import re
# validovat že zadané studenstké číslo je zadané správně (např. F26303, Z01958)
def validate_student_id(student_id):
    pattern = r'^[FZ]\d{5}$'
    return bool(re.match(pattern, student_id))

print(validate_student_id("F26303"))  # True
print(validate_student_id("FF26303"))  # False
print(validate_student_id("Z01958"))  # True
print(validate_student_id("X00000"))  # False
```

```python
%%R

validate_student_id <- function(student_id) {
    pattern <- "^[FZ]\\d{5}$"
    return(grepl(pattern, student_id))
    }

print(validate_student_id("F26303"))  # TRUE
print(validate_student_id("FF26303"))  # FALSE
print(validate_student_id("Z01958"))  # TRUE
```

```python

# jak bychom mohli kontrolvat že je správně zadané jméno: 2 slova, první písmeno velké

def validate_name(name):
    pattern = r'^[A-Z][a-z]+ ([A-Z][a-z]+ )?[A-Z][a-z]+$'
    return bool(re.match(pattern, name)) 

print(validate_name("John Doe"))  # True
print(validate_name("john doe"))  # False
print(validate_name("John"))  # False
print(validate_name("John Doe Smith"))  # True
```

```python
%%R

validate_name <- function(name) {
    pattern <- "^[A-Z][a-z]+ ([A-Z][a-z]+ )?[A-Z][a-z]+$"
    return(grepl(pattern, name))
}

print(validate_name("John Doe"))  # TRUE
print(validate_name("john doe"))  # FALSE
print(validate_name("John"))  # FALSE
print(validate_name("John Doe Smith"))  # TRUE
```
